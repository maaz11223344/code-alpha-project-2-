# CodeAlpha Project 2 — Exploratory Data Analysis (EDA)

E-commerce Orders Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("../data/ecommerce_orders_original.xlsx", sheet_name="Sheet1")
df.head()
```

## 1. Dataset Structure

```python
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData Types:\n", df.dtypes)
print("\nDate range:", df["Date"].min().date(), "to", df["Date"].max().date())
```

## 2. Data Quality

```python
quality = pd.DataFrame({"Missing Values":df.isna().sum(),"Missing %":(df.isna().mean()*100).round(2),"Unique Values":df.nunique()})
print(quality)
print("\nDuplicate rows:", df.duplicated().sum())
```

## 3. Data Cleaning

```python
df["CouponCode"] = df["CouponCode"].fillna("NONE")
df["Date"] = pd.to_datetime(df["Date"])
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.to_period("M").astype(str)
df["DayOfWeek"] = df["Date"].dt.day_name()
df["Quarter"] = df["Date"].dt.to_period("Q").astype(str)
print(df.isna().sum())
```

## 4. Descriptive Statistics

```python
df[["Quantity","UnitPrice","ItemsInCart","TotalPrice"]].describe().round(2)
```

## 5. Product Analysis

```python
product_summary=df.groupby("Product").agg(Orders=("OrderID","count"),Revenue=("TotalPrice","sum"),Average_Order_Value=("TotalPrice","mean"),Average_Quantity=("Quantity","mean")).sort_values("Revenue",ascending=False)
product_summary.round(2)
```

```python
product_summary["Revenue"].sort_values().plot(kind="barh",figsize=(9,5))
plt.title("Revenue by Product"); plt.xlabel("Revenue"); plt.ylabel("Product"); plt.tight_layout(); plt.show()
```

## 6. Order Status Analysis

```python
print(df["OrderStatus"].value_counts())
df["OrderStatus"].value_counts().plot(kind="bar",figsize=(9,5))
plt.title("Orders by Status"); plt.xlabel("Status"); plt.ylabel("Orders"); plt.xticks(rotation=30); plt.tight_layout(); plt.show()
```

## 7. Payment Method Analysis

```python
print(df["PaymentMethod"].value_counts())
df["PaymentMethod"].value_counts().plot(kind="bar",figsize=(9,5))
plt.title("Orders by Payment Method"); plt.xlabel("Payment Method"); plt.ylabel("Orders"); plt.xticks(rotation=25); plt.tight_layout(); plt.show()
```

## 8. Referral Source Analysis

```python
ref=df.groupby("ReferralSource")["TotalPrice"].sum().sort_values(ascending=False)
print(ref.round(2))
ref.sort_values().plot(kind="barh",figsize=(9,5))
plt.title("Revenue by Referral Source"); plt.xlabel("Revenue"); plt.ylabel("Referral Source"); plt.tight_layout(); plt.show()
```

## 9. Time Analysis

```python
monthly=df.groupby("Month")["TotalPrice"].sum()
monthly.plot(figsize=(11,5)); plt.title("Monthly Revenue Trend"); plt.xlabel("Month"); plt.ylabel("Revenue"); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
print("Peak month:",monthly.idxmax(),"Revenue:",round(monthly.max(),2))
```

## 10. Order Value Distribution

```python
plt.figure(figsize=(9,5)); plt.hist(df["TotalPrice"],bins=30)
plt.title("Distribution of Order Values"); plt.xlabel("Total Price"); plt.ylabel("Orders"); plt.tight_layout(); plt.show()
```

## 11. Outlier Detection

```python
q1=df["TotalPrice"].quantile(.25); q3=df["TotalPrice"].quantile(.75); iqr=q3-q1
lower=q1-1.5*iqr; upper=q3+1.5*iqr
outliers=df[(df["TotalPrice"]<lower)|(df["TotalPrice"]>upper)]
print("Q1:",round(q1,2),"Q3:",round(q3,2),"IQR:",round(iqr,2))
print("Lower bound:",round(lower,2),"Upper bound:",round(upper,2))
print("Potential outliers:",len(outliers))
outliers[["OrderID","Date","Product","Quantity","UnitPrice","TotalPrice"]].head(10)
```

## 12. Key Findings and Conclusion

The EDA identifies the dataset structure, data-quality issues, product performance, order-status patterns, payment behavior, referral performance, revenue trends, distribution of order values, and potential outliers. The cleaned dataset is ready for further visualization or dashboard work.
